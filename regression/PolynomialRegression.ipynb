Imports

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

add dataset

```python
data = pd.read_csv('datas.csv')
x = data.iloc[: , 1:-1].values
y = data.iloc[: , -1].values
```

Regression model

```python
from sklearn.linear_model import LinearRegression
lin_reg  = LinearRegression()
lin_reg.fit(x,y)
```

Polynomial dataset and regression model

```python
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree= 4)
x_poly_reg = poly_reg.fit_transform(x)
lin_reg1 = LinearRegression()
lin_reg1.fit(x_poly_reg, y)
```

Visualizing Linear Regression

```python
plt.scatter(x, y, color = 'red')
plt.plot(x , lin_reg.predict(x), color = 'blue')
plt.title('Simple Linear Regression')
plt.xlabel('position')
plt.ylabel('salary')
plt.show()
```

polynomial plot

```python
plt.scatter(x, y, color = 'red')
plt.plot(x, lin_reg1.predict(x_poly_reg))
plt.title('polynomial regression')
plt.xlabel('job level')
plt.ylabel('salary')
plt.show()
```

predicting with linear regression

```python
lin_reg.predict([[6.5]])
```

predicting with ploynomial regression

```python
lin_reg1.predict(poly_reg.fit_transform([[6.5]]))
```
