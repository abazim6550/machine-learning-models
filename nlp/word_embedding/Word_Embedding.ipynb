**Word Embedding Template**



```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

Define the vocabulary size and One Hot Representation

```python
voc_size=1000
one_hot_rep = [one_hot(word, voc_size) for word in sent]
one_hot_rep
```

Word Embedding Representation





```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

word padding

```python
sent_len = 8
embedded_sent = pad_sequences(one_hot_rep, maxlen=sent_len, padding= "pre")
embedded_sent
```

```python
dim = 10
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_len))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_sent)
```
