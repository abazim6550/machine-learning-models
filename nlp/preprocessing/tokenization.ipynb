Tokenization

```python
corpus="""A paragraph is a series of sentences that are organized and coherent,
 and are all related to a single topic.
 Almost every piece of writing you do that is longer than a few sentences should be organized into paragraphs.
"""
```

Sentence tokanization

```python
print(corpus)
from nltk.tokenize import sent_tokenize

```

```python
import nltk

nltk.download('punkt_tab')

documents=sent_tokenize(corpus)
```

```python
print(type(documents))
for sent in documents:
  print(sent)

```

word tokenization

```python
from nltk.tokenize import word_tokenize
```

```python
words = word_tokenize(corpus)
print(words)
```

with punctuation ----- it include ' as token from words like player's

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

treebank tokenization    -------- it include . as token

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer=TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
