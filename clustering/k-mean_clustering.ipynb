Imports

````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
````

import dataset

````python
dataset = pd.read_csv('data.csv')
x = dataset.iloc[:,[3,4]].values
````

find numbers of cluster usong elbow method

```
# This is formatted as code
```



````python
from sklearn.cluster import KMeans
wcss = []
for i in range(1, 11):
  kmens = KMeans(n_clusters=i, init="k-means++", random_state= 35)
  kmens.fit(x)
  wcss.append(kmens.inertia_)
plt.plot(range(1, 11), wcss)
plt.title("elbow plot")
plt.xlabel("no of clusters")
plt.ylabel("wcss")
plt.show()
````

Train k-means model on data

````python
k_means = KMeans(n_clusters= 5, init= "k-means++", random_state=35)
y = k_means.fit_predict(x)
````

Visualize clusters

````python
plt.scatter(x[y == 0,0], x[y == 0,1], s=100, c="red", label="cluster1")
plt.scatter(x[y == 1,0], x[y == 1,1], s=100, c="blue", label="cluster1")
plt.scatter(x[y == 2,0], x[y == 2,1], s=100, c="green", label="cluster1")
plt.scatter(x[y == 3,0], x[y == 3,1], s=100, c="orange", label="cluster1")
plt.scatter(x[y == 4,0], x[y == 4,1], s=100, c="pink", label="cluster1")
plt.scatter(k_means.cluster_centers_[:, 0], k_means.cluster_centers_[:, 1], s = 300, c = 'yellow', label = 'Centroids')
plt.title('Clusters')
plt.xlabel('Salary')
plt.ylabel('spending')
plt.legend()
plt.show()
````
